# Plant Succession ML Project: Schema Walkthrough

This notebook loads the public sample CSVs and prints a few basic summaries. The sample data is intentionally small and sanitized; it demonstrates table shape and validation logic, not scientific results.

```python
from pathlib import Path

import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA = ROOT / "data" / "sample"

sites = pd.read_csv(DATA / "site_registry_sample.csv")
observations = pd.read_csv(DATA / "observations_sample.csv")
events = pd.read_csv(DATA / "disturbance_events_sample.csv")

sites.head()
```

```python
print(f"Sites: {len(sites)}")
print(f"Observations: {len(observations)}")
print(f"Disturbance events: {len(events)}")
```

```python
observations_by_site = (
    observations.groupby("site_id")
    .size()
    .rename("observation_count")
    .reset_index()
    .merge(sites[["site_id", "site_name", "habitat_context"]], on="site_id", how="left")
    .sort_values(["observation_count", "site_id"], ascending=[False, True])
)

observations_by_site
```

```python
events_by_type = (
    events.groupby("disturbance_event_type")
    .size()
    .rename("event_count")
    .reset_index()
    .sort_values("event_count", ascending=False)
)

events_by_type
```

```python
cover_columns = [
    "estimated_total_vegetation_cover_pct",
    "bare_ground_pct",
    "gravel_rock_pct",
    "litter_mulch_woodchip_pct",
    "standing_dead_pct",
]

cover_summary = observations[cover_columns].describe().round(1)
cover_summary
```

```python
plot_ready_cover = observations[["observation_date", "site_id", *cover_columns]].melt(
    id_vars=["observation_date", "site_id"],
    value_vars=cover_columns,
    var_name="cover_metric",
    value_name="percent_cover",
)

plot_ready_cover.head(10)
```
